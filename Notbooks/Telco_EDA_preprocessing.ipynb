```python
df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan).astype(float)

#df['TotalCharges'] = df['TotalCharges'].astype(float)

df['TotalCharges'].isnull().sum()
```

```python
#TotalCharges
df['TotalCharges']

'''#data has one  " " that needs to be placed with NA and then imputed
#df['TotalCharges'] = df['TotalCharges'].replace(" ", np.nan)

#needs to converted into integer
df['TotalCharges'].isnull().sum()'''
```

```python
#total charges has 11 null values that can be imputed with mean

df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].mean())

df['TotalCharges'].dtype
```

Data Cleaning

```python
no_int_service = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection' , 'TechSupport' , 'StreamingTV', 'StreamingMovies']

for i in no_int_service:
    df[i] = df[i].replace('No internet service', 'No')
    

df['StreamingMovies'].value_counts()
```
